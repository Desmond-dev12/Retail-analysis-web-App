## 3. **Exploratory Data Analysis (EDA)**

```python
import plotly.express as px
```

```python
store_sales = data.groupby("Store")["Total"].sum().reset_index().round(2)
fig_store_sales = px.bar(store_sales, x="Store", y="Total",
                         title="Revenue by Store Sales",
                         text= "Total", color= "Store", text_auto=".2s",
                         color_discrete_sequence=px.colors.sequential.Viridis)
fig_store_sales.show()
```

```python
category_sales = data.groupby("Category")["Total"].sum().reset_index().round(2)
fig_category_sales = px.bar(category_sales, x="Category", y="Total",
                         title="Revenue by Category Sales",
                         text= "Total", color= "Category", text_auto=".2s",
                         color_discrete_sequence=px.colors.sequential.PuBu_r)
fig_category_sales.show()
```

```python
avg_category_sales = data.groupby("Category")["Rating"].mean().reset_index().round(2)
fig_avg_category_sales = px.bar(avg_category_sales, x="Category", y="Rating",
                         title="Average Rating by Category",
                         text= "Rating", color= "Category", text_auto=".2s" )

fig_avg_category_sales.show()
```

```python
Customer_sales = data.groupby("Customer Type")["Total"].sum().reset_index().round(2)

fig_customer_sales = px.pie(Customer_sales,
                         names="Customer Type",
                         values="Total",
                         title="Revenue by Customer Type",
                         color="Customer Type",
                         hole=0.4)
fig_customer_sales.show()
```

```python

monthly_sales = data.groupby("Date")["Total"].sum().reset_index().round(2)
fig_monthly_sales = px.line(monthly_sales, x="Date", y="Total", markers=True,
                         title="Daily Sales Trend", color_discrete_sequence=px.colors.sequential.Plasma
                        )
fig_monthly_sales.show()
```

```python
day_order= ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
data["Day"] = pd.Categorical(data["Day"], categories=day_order, ordered=True)
dail_sales_accross_week = data.groupby("Day")["Total"].sum().reset_index().round(2)
fig_dail_sales_accross_week = px.bar(dail_sales_accross_week, x="Day", y="Total",
                         title="Daily Sales Accross Week",
                         text= "Total", color= "Day",
                         text_auto= ".2s")
fig_dail_sales_accross_week.show()
```

## 4. Key Business Insights & Actionable Recommendations

### 🔍 Key Business Insights
1. **Branch Revenue:** West Hills Mall leads total revenue (77k GHS), outperforming Kumasi City Mall (68k GHS) and Accra Mall (67k GHS).
2. **Top Product Lines:** Groceries (41k GHS) and Furniture (40k GHS) are the strongest revenue drivers, while Stationery (27k GHS) lags behind.
3. **Weekly Sales Patterns:** Revenue heavily spikes on Saturday (48k GHS) and Sunday (34k GHS), with an isolated mid-week surge on Wednesday (34k GHS).
4. **Customer Distribution:** Sales revenue is split almost evenly between Members (49.8%) and Normal customers (50.2%).

---

###  Actionable Business Recommendations
1. **Optimize Weekend Staffing & Checkout:** Maximize floor staff and open all checkout lanes on Saturdays and Sundays to handle high foot traffic and prevent cart abandonment.
2. **Launch Mid-Week Promotions:** Introduce targeted Monday and Thursday deals (e.g., discounts on Stationery or double loyalty points) to boost slow trading days.
3. **Buffer High-Demand Inventory:** Maintain higher safety stock levels for Groceries and Furniture ahead of weekends at West Hills Mall to avoid stockouts.
4. **Drive Member Loyalty:** Convert "Normal" shoppers into "Members" at checkout by offering exclusive instant discounts on top categories.
